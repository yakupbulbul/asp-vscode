# ASP Studio in a notebook

Install with `pip install asp-studio[notebook]`, then load the extension. Every `%%asp` cell is solved with clingo and shown with the **Animator**, **Answer Sets**, **Graph** and **Console** tabs.

```python
%load_ext asp_studio
```

## Recursion: step through the fixpoint
Open the *Animator* tab and press **▶** or the step buttons: round 0 holds the facts, every later round adds what the rules can newly derive.

```python
%%asp
% parent(X,Y): X is a parent of Y
parent(anna,bob).
parent(bob,carl). parent(bob,dana).
parent(carl,eve).

ancestor(X,Y) :- parent(X,Y).
ancestor(X,Z) :- parent(X,Y), ancestor(Y,Z).
```

## Choice and constraints: several answer sets
`--tab answers` opens the Answer Sets tab first.

```python
%%asp --tab answers
color(red;green;blue).
node(1..3). edge(1,2). edge(2,3). edge(1,3).

{ c(N,C) : color(C) } = 1 :- node(N).
:- edge(A,B), c(A,C), c(B,C).
#show c/2.
```

## Optimisation
Only optimal models are kept; the cost is shown on the answer card.

```python
%%asp --tab answers
item(a;b;c;d).
weight(a,3). weight(b,5). weight(c,2). weight(d,4).
value(a,4). value(b,7). value(c,2). value(d,5).

{ take(I) : item(I) }.
:- #sum { W,I : take(I), weight(I,W) } > 9.
#maximize { V,I : take(I), value(I,V) }.
#show take/1.
```

## From Python
`solve` returns a `Result` you can inspect; showing it as the last expression of a cell displays the visualizer.

```python
import asp_studio as asp

res = asp.solve('a :- not b. b :- not a.')
print(res.result, [sorted(m) for m in res])
res
```
